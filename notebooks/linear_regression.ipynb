```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")

df = pd.read_csv("../dataset/KAG_energydata_complete.csv")

print("Dataset loaded successfully!")

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

df.head()

df.info()

df.describe()

missing_values = df.isnull().sum()

print(missing_values)

print("Total duplicate records:", df.duplicated().sum())
```

```python
print(df.columns.tolist())
print("Mean Energy Consumption:", df["Appliances"].mean())
print("Maximum Energy Consumption:", df["Appliances"].max())
print("Minimum Energy Consumption:", df["Appliances"].min())
print("Standard Deviation:", df["Appliances"].std())

plt.figure(figsize=(10, 5))

plt.hist(df["Appliances"], bins=30)

plt.title("Distribution of Appliance Energy Consumption")
plt.xlabel("Energy Consumption (Wh)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["T1"], df["Appliances"], alpha=0.3)

plt.title("Kitchen Temperature vs Energy Consumption")
plt.xlabel("Kitchen Temperature (°C)")
plt.ylabel("Energy Consumption (Wh)")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["RH_1"], df["Appliances"], alpha=0.3)

plt.title("Kitchen Humidity vs Energy Consumption")
plt.xlabel("Kitchen Humidity (%)")
plt.ylabel("Energy Consumption (Wh)")

plt.show()
```

```python
correlation = df.corr(numeric_only=True)

print(correlation["Appliances"].sort_values(ascending=False))

plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
df["date"] = pd.to_datetime(df["date"])

df = df.sort_values("date")

plt.figure(figsize=(15, 5))

plt.plot(df["date"], df["Appliances"])

plt.title("Energy Consumption Over Time")
plt.xlabel("Date")
plt.ylabel("Energy Consumption (Wh)")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
target_correlation = correlation["Appliances"].drop("Appliances")

target_correlation = target_correlation.sort_values(
    key=abs,
    ascending=False
)

print(target_correlation)
```
